```python
from google.colab import files
from google.colab.patches import cv2_imshow
import cv2

uploaded = files.upload()
filename = list(uploaded.keys())[0]

img = cv2.imread(filename)
cv2_imshow(img)
```

```python
from deepface import DeepFace
import cv2
```

```python
# Analyze image
results = DeepFace.analyze(
    img_path = filename,
    actions = ['age', 'emotion'],
    detector_backend = 'retinaface'
)
```

```python
# Extract values
expression = max(results[0]['emotion'], key=results[0]['emotion'].get)
age = results[0]['age']
```

```python
music = {
    "angry": ["Stress Relief Frequency", "Peaceful Sleep"],
    "happy": ["Happy Vibes", "Feel Good Songs"],
    "sad": ["Calm Piano", "Healing Music"],
    "neutral": ["Soft Instrumental"],
    "fear": ["Relaxing Nature Sounds"]
}

# Analyze image to define 'results' (moved from w9Hc8AD2wZRJ to ensure 'results' is available)
results = DeepFace.analyze(
    img_path = filename,
    actions = ['age', 'emotion'],
    detector_backend = 'retinaface'
)

# Define 'expression' from the analysis results
expression = max(results[0]['emotion'], key=results[0]['emotion'].get)

print("\n🎧 Music Recommendation:")
for song in music.get(expression, ["Relaxing Music"]):
    print("-", song)
```

```python
# Convert to age range (example: 27 -> 24-30)
lower = age - 3
upper = age + 3
age_range = f"{lower}-{upper}"
```

```python
print("\n--- RESULT ---")
print(f"Expression: {expression}")
print(f"Estimated Age: {age_range}")
```
